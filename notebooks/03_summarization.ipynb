```python
import time, warnings
warnings.filterwarnings("ignore")
from transformers import pipeline, logging
logging.set_verbosity_error()
summarizer = pipeline("summarization", model="facebook/bart-large-cnn", device="cpu")
```

```python
# 주장, 반론, 결론이 뒤에 오는 논설문 (직접 작성)
text4 = ("Large language models (LLMs) such as GPT-4 and Claude have rapidly moved from research labs into everyday "
"products, but their adoption raises a tension between convenience and reliability. On one hand, many office "
"workers and students now use a chatbot every week, mostly for drafting emails, summarizing documents, and writing code. Proponents argue that natural-language interfaces lower the barrier to "
"computing, because users no longer need to learn menus or query languages. On the other hand, critics point out "
"that these models sometimes 'hallucinate', producing fluent statements that are factually wrong, and that users "
"tend to trust confident text even when it is incorrect. A lawyer in New York, for example, was sanctioned after "
"submitting a brief that cited court cases invented by a chatbot. Researchers in human-computer interaction therefore "
"suggest design changes rather than simply better models: showing sources next to answers, expressing uncertainty, "
"and keeping a human in the loop for high-stakes decisions. Privacy is a further concern, since prompts may contain "
"personal or confidential data that is stored by the provider. Whether LLMs become a trusted everyday interface may "
"depend less on raw accuracy than on whether people can understand when, and how much, to rely on them.")
```

```python
for name in ["text", "text1", "text2", "text3", "text4"]:
    src = globals()[name]
    t = time.time()
    out = summarizer(src, max_length=100, min_length=30, do_sample=False)[0]["summary_text"]
    print(f"===== {name}  ({len(src.split())} words -> {len(out.split())} words, {time.time()-t:.1f}s)")
    print(out, "\n")
```
